# Simulação de Monte Carlo

A Simulação de Monte Carlo é um método utilizado para prever e compreender o comportamento de sistemas que envolvem incertezas. Ao executar múltiplas simulações com entradas aleatórias, essa técnica ajuda a estimar possíveis resultados e suas probabilidades.

Em vez de depender de uma única solução, ela gera diversos cenários diferentes para fornecer uma visão mais clara da gama de resultados possíveis. Imagine lançar uma moeda diversas vezes para descobrir a probabilidade de dar cara ou coroa; quanto mais lançamos, mais clara se torna a probabilidade. É uma técnica utilizada em áreas como finanças, engenharia e ciência.

A Simulação de Monte Carlo funciona seguindo estes passos fundamentais:

* Definição do Problema: Identificar o sistema ou processo que desejamos simular. Pode ser qualquer coisa, desde preços de ações até modelos de engenharia.

* Modelagem das Entradas: Escolher as distribuições de probabilidade adequadas para os parâmetros incertos do modelo. Por exemplo, ao modelar preços de ações, podemos usar uma distribuição normal para representar as flutuações de preço.

* Execução das Simulações: Gerar entradas aleatórias com base nas distribuições escolhidas e repetir o processo muitas vezes (geralmente milhares ou milhões) para criar uma variedade de cenários possíveis.

* Análise dos Resultados: Após rodar as simulações, obtemos uma distribuição de resultados. Ao analisar esses dados, podemos tomar decisões fundamentadas com base na gama de cenários possíveis, e não apenas em um único resultado esperado.

O objetivo principal da simulação de Monte Carlo é usar amostragem aleatória para estimar o valor esperado de uma função $f$ sobre um domínio $D$. Dada uma função $f(x)$ e um domínio $D$, o valor esperado $\mathbb{E}[f(x)]$ pode ser estimado usando a seguinte fórmula (pela lei dos grandes números):
$$
    \mathbb{E}[f(x)] \approx \frac{1}{N} \sum_{i=1}^{N} f(x_i)
$$

Monte Carlo also helps estimate integrals when the exact solution is hard to find. For that, the formula is:
$$
    \int_D f(x) \, dx \approx \frac{1}{N} \sum_{i=1}^{N} f(x_i) \cdot \text{volume}(D)
$$

## Estimando $\pi$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def monte_carlo_pi(num_samples):
    x = np.random.uniform(-1, 1, num_samples)
    y = np.random.uniform(-1, 1, num_samples)
    
    distance = np.sqrt(x**2 + y**2)
    
    inside_circle = np.sum(distance <= 1)
    
    pi_estimate = (inside_circle / num_samples) * 4
    
    return pi_estimate
```

```python
num_samples = 100000

pi_estimate = monte_carlo_pi(num_samples)
```

```python
print(f"Estimated value of π with {num_samples} samples: {pi_estimate}")
x = np.random.uniform(-1, 1, num_samples)
y = np.random.uniform(-1, 1, num_samples)
plt.figure(figsize=(6, 6))
plt.scatter(x, y, c=np.sqrt(x**2 + y**2) <= 1, cmap='bwr', s=1)
plt.gca().set_aspect('equal', adjustable='box')
plt.title('Monte Carlo Simulation for π')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```
